Task 01: Create a Bar Chart or Histogram

The task is to visualize the distribution of a categorical variable (like Gender) or a continuous variable (like Age).


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import pandas as pd

df = pd.read_csv(
    "/content/API_SP.POP.TOTL_DS2_en_csv_v2_33112.csv",
    skiprows=4
)

print(df.head())
```

```python
df.head()
```

```python
df.columns
```

```python
df[['Country Name', '2025']].head()
```

```python
top10 = df.nlargest(10, '2025')

top10[['Country Name', '2025']]
```

```python
plt.figure(figsize=(10, 6))

plt.bar(top10['Country Name'], top10['2025'])

plt.xlabel('Country')
plt.ylabel('Population')
plt.title('Top 10 Countries by Population in 2025')

plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(top10['Country Name'], top10['2025'])

plt.xlabel('Country')
plt.ylabel('Population')
plt.title('Top 10 Countries by Population in 2025')

plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.savefig('task01_population_bar_chart.png', dpi=300)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df['2025'].dropna(), bins=20)

plt.xlabel('Population')
plt.ylabel('Number of Countries')
plt.title('Distribution of Population in 2025')

plt.show()
```

```python
import numpy as np

population = df['2025'].dropna()

plt.figure(figsize=(10, 6))

plt.hist(np.log10(population), bins=20)

plt.xlabel('Log10(Population)')
plt.ylabel('Number of Countries/Regions')
plt.title('Distribution of Population in 2025')

plt.show()
```
